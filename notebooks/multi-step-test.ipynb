```python
from dotenv import load_dotenv
load_dotenv()
from ecmwfapi import ECMWFService


def make_cost_check_request(request: dict) -> str:
    """ Create a cost check request string from a request dict. """
    parts = [
        f"{k} = {'/'.join(str(x) for x in v) if k == 'param' and isinstance(v, (list, tuple)) else v}"
        for k, v in request.items()
        if k != "format"
    ]

    cost_check_request = "list, \n"
    cost_check_request += ", ".join(parts)
    cost_check_request += ", output = cost"

    return cost_check_request

Request_0 = {
 'area': '52.5/-1.0/52.300000000000004/-0.9',
 'class': 'od',
 'date': '2016-01-21',
 'format': 'grib2',
 'grid': '0.1/0.1',
 'levtype': 'sfc',
 'param': ['2d',
            '2t',
            '10u',
            '10v',
            '100u',
            '100v',
            'u10n',
            'v10n',
            'i10fg',
            'msl',
            'tp'],
 'step': '0/to/72/by/1',
 'stream': 'oper',
 'time': '12',
 'type': 'fc'
 }
```

```python
service = ECMWFService("mars")
cost_check_request_str = make_cost_check_request(Request_0)
cost = service.execute(req = cost_check_request_str, target='./result/cost_check_req0.txt')
print(f"RESPONSE Request_0:\n\n {cost} \n\nEND OF RESPONSE")
```

```python
import xarray as xr
import pandas as pd

def load_grib_to_dataframe(path: str, var: str | None = None, dropna: bool = True) -> pd.DataFrame:
    """
    Load a .grib2 file into a pandas DataFrame using xarray (cfgrib engine).
    - path: path to the .grib2 file
    - var: optional variable name to extract (e.g. 'tp', 'msl', '2t'); if None, loads all vars
    - dropna: drop rows that are all-NaN (useful for sparse datasets)
    Returns a DataFrame with coordinate columns (time, latitude, longitude, level, ...) and data columns.
    """
    # Requires cfgrib / eccodes installed (xarray will use engine='cfgrib')
    ds = xr.open_dataset(path, engine="cfgrib")
    
    if var is not None:
        if var not in ds:
            raise KeyError(f"Variable '{var}' not found in dataset. Available: {list(ds.variables)}")
        data = ds[var].to_dataframe().reset_index()
    else:
        data = ds.to_dataframe().reset_index()
    
    if dropna:
        # drop rows where all data columns (excluding coords) are NaN
        coord_cols = [c for c in data.columns if c in ("time", "valid_time", "step", "level", "latitude", "longitude")]
        data = data.dropna(how="all", subset=[c for c in data.columns if c not in coord_cols])
    
    return data

# Example usage (uses the filename from this notebook):
grib_path = "./result/ecmwf_data_req0_multi_issue_date.grib2"
df = load_grib_to_dataframe(grib_path)
df.head()
```
