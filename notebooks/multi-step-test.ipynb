```python
from dotenv import load_dotenv
load_dotenv()
from ecmwfapi import ECMWFService


def make_cost_check_request(request: dict) -> str:
    """ Create a cost check request string from a request dict. """
    parts = [
        f"{k} = {'/'.join(str(x) for x in v) if k == 'param' and isinstance(v, (list, tuple)) else v}"
        for k, v in request.items()
        if k != "format"
    ]

    cost_check_request = "list, \n"
    cost_check_request += ", ".join(parts)
    cost_check_request += ", output = cost"

    return cost_check_request

Request_0 = {
 'area': '52.5/-1.0/52.300000000000004/-0.9',
 'class': 'od',
 'date': '2016-01-21',
 'format': 'grib2',
 'grid': '0.1/0.1',
 'levtype': 'sfc',
 'param': ['2d',
            '2t',
            '10u',
            '10v',
            '100u',
            '100v',
            'u10n',
            'v10n',
            'i10fg',
            'msl',
            'tp'],
 'step': '0/to/72/by/1',
 'stream': 'oper',
 'time': '12',
 'type': 'fc'
 }
```

```python
service = ECMWFService("mars")
cost_check_request_str = make_cost_check_request(Request_0)
cost = service.execute(req = cost_check_request_str, target='./result/cost_check_req0.txt')
print(f"RESPONSE Request_0:\n\n {cost} \n\nEND OF RESPONSE")
```
